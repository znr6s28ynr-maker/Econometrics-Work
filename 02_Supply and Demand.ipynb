```python
#importing libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
import statsmodels.api as sm
```

```python
#creating quantities
quantity = np.arange(5, 9, .5)

#creating the demand curve:
demand_latte_price = 10 - 0.5 * quantity

#plotting the curve
plt.figure(figsize=(8, 8))
plt.plot(quantity, demand_latte_price, color='blue', label='Demand Curve')
plt.title('Demand Curve for Tall Starbucks Lattes in North America')
plt.xlabel('Quantity Demanded Per Day (in millions of units)')
plt.ylabel('Price ($)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
#creating quantities
quantity = np.arange(5, 9, .5)

#creating the supply curve:
supply_latte_price = 3.25 + 0.5 * quantity

#plotting the curve
plt.figure(figsize=(8, 8))
plt.plot(quantity, supply_latte_price, color='red', label='Supply Curve')
plt.title('Supply Curve for Tall Starbucks Lattes in North America')
plt.xlabel('Quantity Supplied Per Day (in millions of units)')
plt.ylabel('Price ($)')
plt.legend()
plt.grid(True)
plt.show()
```

These are fictional supply and demand curves that I tried to center around the general variation of a tall Starbucks latte price and the number of tall Starbucks lattes ordered every day in North America. As the price rises, the demand curve shows that the quantity of lattes demanded will go down. Conversely, the supply curve shows that as the price of a latte increases, the quantity supplied will rise. Equilibrium would be where the two curves intersect.

The supply and demand curves could shift for several reasons:

For instance, if the cost of coffee beans suddenly rose, the number of lattes Starbucks would be willing to supply at any given price would go down, sending the supply curve down and to the right.

If it suddenly became really cool for highschoolers to walk into class with a tall Starbucks latte, then the demand curve might shift upward because the quantity of lattes demanded at a given price would rise.

```python
#creating quantities
quantity = np.arange(5, 9, .5)

#original curve slopes
latte_demand_slope = -0.5
latte_supply_slope = 0.5

#plotting original curves
plt.figure(figsize=(8, 8))

#list of equilibrium points
equilibrium_points = []

#creating scale of shift
for i in range(5):
  latte_demand_shift = 0.25 * i * (-1)**i
  latte_supply_shift = 0.5 * i * (-1)**i

  #defining shift intercepts
  latte_demand_intercept = 9 + latte_demand_shift
  latte_supply_intercept = 2 + latte_supply_shift

  #creating shifted demand curves
  latte_demand_price = latte_demand_intercept + latte_demand_slope * quantity

  #creating shifted supply curves
  latte_supply_price = latte_supply_intercept + latte_supply_slope * quantity

  #plotting demand curves
  plt.plot(quantity, latte_demand_price, color='blue', label=f'Demand Curve {i+1}' if i == 0 else "_nolegend_")

  #plotting supply curves
  plt.plot(quantity, latte_supply_price, color='red', label=f'Supply Curve {i+1}' if i == 0 else "_nolegend_")

  #caculating points of intersection (equilibrium)
  equilibrium_quantity = (latte_demand_intercept - latte_supply_intercept) / (latte_supply_slope - latte_demand_slope)
  equilibrium_price = latte_demand_intercept + latte_demand_slope * equilibrium_quantity

  #storing intersectino points
  equilibrium_points.append((equilibrium_quantity, equilibrium_price))

#dressing up the graph
plt.title('Market Shifts For Tall Starbucks Lattes')
plt.xlabel('Quantity (in millions of units)')
plt.ylabel('Price ($)')
plt.grid(True)
plt.show()
```

```python
#extracting equilibrium points
equilibrium_quantities = np.array([point[0] for point in equilibrium_points]).reshape(-1, 1)
equilibrium_prices = np.array([point[1] for point in equilibrium_points])

#fitting linear regression model
model = LinearRegression().fit(equilibrium_quantities, equilibrium_prices)

#creating predicted prices along line
predicted_prices = model.predict(equilibrium_quantities)

#extracting slope of line
slope = model.coef_[0]

#plotting points of equilibrium
plt.figure(figsize=(8, 8))
plt.scatter(equilibrium_quantities, equilibrium_prices, color='purple', label='Equilibrium Points')

#plotting line through equilibrium points
plt.plot(equilibrium_quantities, predicted_prices, color='green', label='Line of Best Fit')

#labeling the slope on graph
plt.text(equilibrium_quantities.min() + 0.5, equilibrium_prices.max() - 1, f'Slope: {slope:.2f}', fontsize=12, color='green')

#dressing the graph
plt.title('Equilibrium Points with Line of Best Fit')
plt.xlabel('Equilibrium Quantity (in millions of units)')
plt.ylabel('Equilibrium Price ($)')
plt.legend(loc='lower right')
plt.grid(True)
plt.show()
```

From what I understand, plotting points of equilibrium on a graph and then creating a linear regression model to connect the points is in many ways the same as plotting recorded prices against the quantity and then hoping to derive the demand or supply curve.

But this way of thinking is erroneous, since each point doesn't represent a point along a single stable supply or demand curve. Rather, each point represents the intersection of a supply and demand curve representative of a specific market situation.

There are so many external factors shifting the theoretical demand and supply curves at any given time that we can't conclude that the line in some way identifies the supply or demand curve.

```python
#creating quantities
quantity = np.arange(5, 9, .5)

#original curve slopes
latte_demand_slope = -0.5
latte_supply_slope = 0.5

#plotting original curves
plt.figure(figsize=(8, 8))

#clearing list of equilibrium points
equilibrium_points = []

#creating the demand curve
demand_latte_price = 10 - 0.5 * quantity

#plotting demand curve
plt.plot(quantity, demand_latte_price, color='blue')

#creating demand intercept
latte_demand_intercept = 10

#creating supply shift
for i in range(10):
  latte_supply_shift = 0.5 * i * (-1)**i

  #defining supply shift intercepts
  latte_supply_intercept = 2 + latte_supply_shift

  #creating shifted supply curves
  latte_supply_price = latte_supply_intercept + latte_supply_slope * quantity

  #plotting supply curves
  plt.plot(quantity, latte_supply_price, color='red', label=f'Supply Curve {i+1}' if i == 0 else "_nolegend_")

  #caculating points of intersection (equilibrium)
  equilibrium_quantity = (latte_demand_intercept - latte_supply_intercept) / (latte_supply_slope - latte_demand_slope)
  equilibrium_price = latte_demand_intercept + latte_demand_slope * equilibrium_quantity

  #storing intersectino points
  equilibrium_points.append((equilibrium_quantity, equilibrium_price))

#dressing up the graph
plt.title('Supply Market Shifts For Tall Starbucks Lattes')
plt.xlabel('Quantity (in millions of units)')
plt.ylabel('Price ($)')
plt.grid(True)
plt.show()
```

```python
#extracting equilibrium points
equilibrium_quantities = np.array([point[0] for point in equilibrium_points]).reshape(-1, 1)
equilibrium_prices = np.array([point[1] for point in equilibrium_points])

#fitting linear regression model
model = LinearRegression().fit(equilibrium_quantities, equilibrium_prices)

#creating predicted prices along line
predicted_prices = model.predict(equilibrium_quantities)

#extracting slope of line
slope = model.coef_[0]

#plotting points of equilibrium
plt.figure(figsize=(8, 8))
plt.scatter(equilibrium_quantities, equilibrium_prices, color='purple', label='Equilibrium Points')

#plotting line through equilibrium points
plt.plot(equilibrium_quantities, predicted_prices, color='green', label='Line of Best Fit')

#labeling the slope on graph
plt.text(equilibrium_quantities.min() + 3.5, equilibrium_prices.max() - 1.5, f'Slope: {slope:.2f}', fontsize=12, color='green')

#dressing the graph
plt.title('Equilibrium Points Along Stable Demand Curve')
plt.xlabel('Equilibrium Quantity (in millions of units)')
plt.ylabel('Equilibrium Price ($)')
plt.legend(loc='lower right')
plt.grid(True)
plt.show()
```

Because these equilibrium points were all extracted from the same demand curve, I think it would be reasonable to conclude that they represent the demand curve. The slope of this line is the same as the original demand curve because it IS the original demand curve. Because only the supply curve is shifting, it allows us to see what might really be going on with the demand curve.

If we wanted to be able to isolate either the demand or supply curve in this example, we would need more variables. We'd need data on other factors that might shift these curves, but they'd have to be factors that only shift one or the other, because otherwise, you can't really make any assumptions about the supply/demand curve intersection that produced the market situation in question.

In my Starbucks latte example, if we wanted to understand the supply curve, we would need data on factors that only directly influence demand, such as:


*   what else is currently on the menu (people might come in for a new item, for instance and end up also buying a latte at the same time.)
*   Assuming the lattes in my example are hot, maybe temperature data at time of purchase, since hot weather might drive customers toward cold drinks
*   The price of a similar drink at coffee shops in the vicinity

If we wanted to understand the demand curve, we would then examine parameters that only directly influence supply, like:


*   Cost of the coffee beans Starbucks uses to make lattes
*   Cost of maintaining the Starbucks store
*   Data on staffing (do they have enough employees to keep up with demand?)
